```python
import subprocess
from pathlib import Path



def generar_modelo_datos(mermaid: str) -> str:

    archivo_mmd = Path("archivo.mmd")
    archivo_png = Path("modelo_datos.png")

    archivo_mmd.write_text(
        mermaid,
        encoding="utf-8"
    )

    subprocess.run(
        [
            "mmdc",
            "-i",
            str(archivo_mmd),
            "-o",
            str(archivo_png),
        ],
        check=True
    )

    return f"Modelo de datos generado en: {archivo_png.absolute()}"
```

```python
generar_modelo_datos("""
graph TD
    A[Start] --> B{Decision}
    B -->|Yes| C[Action 1]
    B -->|No| D[Action 2]
    C --> E[End]
    D --> E

""")
```

```python
import base64
import io
import requests
from PIL import Image as im

# Tu diagrama en sintaxis Mermaid
codigo_mermaid = """
graph TD
    A[Inicio] --> B{¿Funciona?}
    B -- Sí --> C[¡Genial!]
    B -- No --> D[Revisar código]
"""

# Codificar el diagrama a base64 seguro para URL
graphbytes = codigo_mermaid.encode("utf8")
base64_bytes = base64.urlsafe_b64encode(graphbytes)
base64_string = base64_bytes.decode("ascii")

# Obtener la imagen desde la API de Mermaid.ink
url = f"https://mermaid.ink/img/{base64_string}"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    # Abrir y guardar la imagen localmente
    imagen = im.open(io.BytesIO(respuesta.content))
    imagen.save("diagrama.png")
    print("¡Imagen guardada exitosamente como diagrama.png!")
else:
    print("Error al generar la imagen.")
```
