```python
import io
class CapturaConsola(io.StringIO):
    def __init__(self):
        super().__init__()
        self.callbacks = []
        self.stdout_original = sys.stdout
        self.stderr_original = sys.stderr

    def write(self, s):
        super().write(s)
        try:
            self.stdout_original.write(s)
        except Exception:
            pass
        for callback in self.callbacks:
            try:
                callback(s)
            except Exception:
                pass

    def flush(self):
        super().flush()
        try:
            self.stdout_original.flush()
        except Exception:
            pass

    def add_callback(self, callback):
        if callback not in self.callbacks:
            self.callbacks.append(callback)

    def remove_callback(self, callback):
        if callback in self.callbacks:
            self.callbacks.remove(callback)
```

```python
import sys

captura_consola = CapturaConsola()
sys.stdout = captura_consola
sys.stderr = captura_consola
```

## Con subprocess
LA idea es probar con subprocess para identificar lo que me manda la cmd

```python
import subprocess

# Ejecutamos el comando 'dir' (equivalente a listar archivos en Windows)
resultado = subprocess.run(["dir"], capture_output=True, text=True, shell=True)
```

```python
# Mostramos la salida estándar (lo que muestra la consola)
print("Salida:")
print(resultado.stdout)
```

Por medio de sys

```python
import sys

# Save original stdout
o = sys.stdout

# Redirect stdout to a file
with open('output.txt', 'w') as f:
    sys.stdout = f
    for i in range(10):
        print("printing line", i)

# Restore stdout
sys.stdout = o
```
